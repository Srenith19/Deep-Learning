```python
import tensorflow as tf, numpy as np

path = tf.keras.utils.get_file('shakespeare.txt',
    'https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt')
text = open(path).read().lower()
chars = sorted(set(text))
c2i = {c: i for i, c in enumerate(chars)}
i2c = {i: c for c, i in c2i.items()}

seq_len = 40
X = np.array([[c2i[c] for c in text[i:i+seq_len]] for i in range(len(text) - seq_len)])
y = np.array([c2i[text[i+seq_len]] for i in range(len(text) - seq_len)])

model = tf.keras.Sequential([
    tf.keras.layers.Embedding(len(chars), 64),
    tf.keras.layers.LSTM(128),
    tf.keras.layers.Dense(len(chars), activation='softmax')])
model.compile(loss='sparse_categorical_crossentropy', optimizer='adam')
model.fit(X, y, batch_size=128, epochs=1)

def generate(seed, length=300):
    seq = [c2i[c] for c in seed.lower()]
    for _ in range(length):
        p = model.predict(np.array(seq[-seq_len:]).reshape(1, -1), verbose=0)[0]
        seq.append(np.random.choice(len(p), p=p))
    return ''.join(i2c[i] for i in seq)

print(generate("shall i compare thee to a summer's day?\n"))
```
