```python
import os, sys, subprocess, urllib.request, itertools, time
import torch
from torchvision.utils import save_image
from torchmetrics.image.inception import InceptionScore
from datasets import load_dataset

t0 = time.time()
def log(msg): print(f"[{time.time()-t0:6.1f}s] {msg}")

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
log(f"Using device: {device}")        # must say cuda
NUM_IMAGES = 30

# 1. Clone repo + download generator (skipped if already present)
if not os.path.exists('stylegan2-ada-pytorch'):
    subprocess.run(['git', 'clone', '--depth', '1',
                    'https://github.com/NVlabs/stylegan2-ada-pytorch.git'])
sys.path.append(os.path.abspath('stylegan2-ada-pytorch'))
if not os.path.exists('ffhq.pkl'):
    urllib.request.urlretrieve(
        'https://nvlabs-fi-cdn.nvidia.com/stylegan2-ada-pytorch/pretrained/ffhq.pkl', 'ffhq.pkl')
log("repo + model ready")

# 2. Generate fakes
import legacy, dnnlib
with open('ffhq.pkl', 'rb') as f:
    G = legacy.load_network_pkl(f)['G_ema'].to(device)
os.makedirs('generated_images', exist_ok=True)
c = torch.zeros([1, G.c_dim], device=device)
fakes = []
with torch.no_grad():
    for i in range(NUM_IMAGES):
        z = torch.randn([1, G.z_dim], device=device)
        img = ((G(z, c, truncation_psi=0.5, noise_mode='const') + 1) * 0.5).clamp(0, 1)
        save_image(img, f'generated_images/fake_{i}.png')
        fakes.append(img.cpu())
fakes = torch.cat(fakes)
log(f"generated {len(fakes)} images")

# 3. Inception Score
is_metric = InceptionScore(splits=5)
is_metric.update((fakes * 255).to(torch.uint8))
mean, std = is_metric.compute()
log(f"Inception Score: {mean:.2f} ± {std:.2f}")

# 4. Real images (small size = faster)
os.makedirs('real_images', exist_ok=True)
stream = load_dataset("nielsr/CelebA-faces", split="train", streaming=True)
for i, item in enumerate(itertools.islice(stream, NUM_IMAGES)):
    item["image"].convert("RGB").resize((256, 256)).save(f"real_images/real_{i}.png")
log("real images saved")

# 5. FID (generated images are 1024x1024; pytorch_fid resizes both internally)
subprocess.run(["python", "-m", "pytorch_fid", "real_images", "generated_images",
                "--dims", "64", "--device", str(device)])
log("FID done")

# 6. Visual comparison
import matplotlib.pyplot as plt
from PIL import Image
fig, axs = plt.subplots(2, 4, figsize=(12, 6))
for k in range(4):
    axs[0, k].imshow(Image.open(f'real_images/real_{k}.png')); axs[0, k].axis('off')
    axs[1, k].imshow(Image.open(f'generated_images/fake_{k}.png')); axs[1, k].axis('off')
axs[0, 0].set_title("Real (top)"); axs[1, 0].set_title("Generated (bottom)")
plt.tight_layout(); plt.show()
```
